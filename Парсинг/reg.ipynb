```python
import os
import time
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
```

```python
def captcha(driver, wait, image_path):
    #Открываем новую вкладку и переключаемся на неё
    driver.execute_script("window.open('');")
    driver.switch_to.window(driver.window_handles[-1])
    driver.get("https://www.prepostseo.com/ru/image-to-text")


    #Ожидаем загрузки поля для отправки файла и передаем путь к картинке
    file_input = wait.until(EC.presence_of_element_located(( By.XPATH, '//input[@type="file"]')))
    absolute_path = os.path.abspath(image_path)
    time.sleep(15)
    file_input.send_keys(absolute_path)

    #Находим кнопку отправки и нажимаем
    knopka = wait.until(EC.element_to_be_clickable((By.XPATH, "/html/body/main/section[1]/div[4]/div[3]/div[1]")))
    knopka.click()
    time.sleep(10)

    text = driver.find_element(By.XPATH, '//*[@id="result__text0"]').text
    
    #Закрываем вкладку и возвращаемся на исходную страницу
    driver.close()
    driver.switch_to.window(driver.window_handles[0])
    
    return text
```

```python
print('Ваша информация для входа')
name=input('Введите ваше имя:')
surname=input('Введите вашу фамилию:')
email=input('Введите вашу почту:')
passvord=input('Введите пароль(делайте его надежным, не менее 8 символов, используя цифры):')

driver = webdriver.Chrome()
driver.get('https://lessons.zennolab.com/ru/advanced')

wait = WebDriverWait(driver, 15)
login_name = wait.until(EC.presence_of_element_located((By.XPATH, '//*[@id="fname"]')))
login_name.clear()
login_name.send_keys(name)

login_surname = wait.until(EC.presence_of_element_located((By.XPATH, '//*[@id="lname"]')))
login_surname.clear()
login_surname.send_keys(surname)

login_email = wait.until(EC.presence_of_element_located((By.XPATH, '//*[@id="email"]')))
login_email.clear()
login_email.send_keys(email)

login_passvord = wait.until(EC.presence_of_element_located((By.XPATH, '//*[@id="pass1"]')))
login_passvord.clear()
login_passvord.send_keys(passvord)

login_passvord2 = wait.until(EC.presence_of_element_located((By.XPATH, '//*[@id="pass2"]')))
login_passvord2.clear()
login_passvord2.send_keys(passvord)

captcha_element = wait.until(EC.visibility_of_element_located(( By.XPATH, '//*[@id="capcha"]')))
captcha_element.screenshot('capcha.png')
print("Скриншот капчи есть")

print("Отправка изображения на распознавание")
text_result = captcha(driver, wait, 'capcha.png')
print(f"Текст получен: '{text_result}'")

login_text=wait.until(EC.presence_of_element_located((By.XPATH, '//*[@id="captcha"]')))
login_text.send_keys(text_result)

button = wait.until(EC.element_to_be_clickable((By.XPATH, '//*[@id="form1"]/input[7]')))
button.click()
print("Капча пройдена")
```
